```r
library(ggplot2)
library(repr)
options(repr.plot.width=9, repr.plot.height=6)
```

```r
data(iris)
ggplot(iris, aes(x = Petal.Length, y = Petal.Width, color = Species)) +
  geom_point(size = 2)
```

```r
irispca <- prcomp(iris[, -5])
summary(irispca)
```

```r
plot(irispca, type = "lines")
```

```r
var <- irispca$sdev^2
pve <- var / sum(var)
S   <- cumsum(pve)
S
```

```r
scores <- data.frame(irispca$x, Species = iris$Species)

ggplot(scores, aes(x = PC1, y = PC2, color = Species)) +
  geom_point(size = 2) +
  labs(title = "PC1 vs PC2")
```
