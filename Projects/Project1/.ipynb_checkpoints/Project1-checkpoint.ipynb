# Project 1

(1)

The data set contains 150 listings for a single used car model. *(Add the source and how the cars were chosen here.)* For each car, `age` is in years (this year's model = 1), `miles` is in thousands of miles and `price` is in thousands of dollars. The goal is to model `price` (response) as a linear function of `age` (predictor).

```r
cars <- read.csv("used_cars_age_miles_price.csv")
str(cars)
summary(cars)
```

(2)

Let x = age and y = price. The commands below compute x̄, s_x, ȳ, s_y, r, SSX, SSXY, SSY (= SSTotal), SSModel and SSE directly.

```r
x <- cars$age
y <- cars$price
n <- length(x)

xbar <- mean(x)
sx   <- sd(x)
ybar <- mean(y)
sy   <- sd(y)
r    <- cor(x, y)

SSX     <- sum((x - xbar)^2)
SSXY    <- sum((x - xbar) * (y - ybar))
SSY     <- sum((y - ybar)^2)   # SSTotal
SSModel <- SSXY^2 / SSX
SSE     <- SSY - SSModel

c(n = n, xbar = xbar, sx = sx, ybar = ybar, sy = sy, r = r)
c(SSX = SSX, SSXY = SSXY, SSY = SSY, SSModel = SSModel, SSE = SSE)
```

(3)

```r
b1 <- SSXY / SSX
b0 <- ybar - b1 * xbar
c(b0 = b0, b1 = b1)

# confirm with lm()
lm.fit <- lm(price ~ age, data = cars)
coef(lm.fit)
```

*Interpretation:* for each additional year of age, the predicted price of this car model drops by about 1.199 thousand dollars (≈ $1,200). The slope is negative because cars depreciate: older cars have more wear and fewer years of use left, so they sell for less. The intercept says a car of age 0 is predicted to cost about $18,092.

(4)

```r
sigma.hat <- sqrt(SSE / (n - 2))
sigma.hat
summary(lm.fit)$sigma   # RSE from lm(), for comparison
```

(5)

```r
plot(cars$age, cars$price, pch = 19, col = "steelblue",
     xlab = "Age (years)", ylab = "Price ($1000s)", main = "Price vs. Age")
abline(lm.fit, col = "red", lwd = 2)
```

(6)

Following ISLR, we plot residuals against fitted values, plus a histogram and a normal Q-Q plot of the residuals (and residuals against age).

```r
par(mfrow = c(2, 2))
plot(fitted(lm.fit), residuals(lm.fit), pch = 19,
     xlab = "Fitted values", ylab = "Residuals", main = "Residuals vs. Fitted")
abline(h = 0, lty = 2)
plot(cars$age, residuals(lm.fit), pch = 19,
     xlab = "Age", ylab = "Residuals", main = "Residuals vs. Age")
abline(h = 0, lty = 2)
hist(residuals(lm.fit), main = "Histogram of Residuals", xlab = "Residuals")
qqnorm(residuals(lm.fit), pch = 19); qqline(residuals(lm.fit))
par(mfrow = c(1, 1))
```

*Linearity:* the residuals are centered on zero but show a mild curve (positive at the smallest and largest fitted values, negative in the middle), so depreciation slows somewhat with age; linearity is a reasonable but imperfect fit.
*Constant variance:* the spread is fairly even across ages, a bit larger for the newest cars (a few very expensive 2-year-old listings). Ages are whole numbers so the points fall in vertical columns.
*Normality:* the histogram is roughly bell-shaped, but the Q-Q plot shows a heavier right tail (a few large positive residuals), so normality is only approximate.
*Independence:* we have no reason to suspect dependence among listings, although this cannot be checked from the data.
Overall the conditions are reasonably met, with mild departures; with n = 150 the inference is fairly robust.

(7)

Following ISLR, we flag a point as an outlier if its studentized residual is beyond ±3, and as high leverage if its leverage h_i is much larger than the average leverage (p+1)/n = 2/150 ≈ 0.013 (we use twice the average, 0.027). For Cook's distance we use the common cutoff D > 1 (D > 0.5 is worth a look).

```r
i <- which.max(abs(residuals(lm.fit)))
cars[i, ]
c(residual    = residuals(lm.fit)[i],
  studentized = rstudent(lm.fit)[i],
  leverage    = hatvalues(lm.fit)[i],
  cooks       = cooks.distance(lm.fit)[i])
mean(hatvalues(lm.fit))   # average leverage, (p+1)/n
2 * mean(hatvalues(lm.fit))
```

The largest residual is for the 2-year-old car with 7.4 thousand miles listed at $22,000 (residual ≈ +6.31, about $6,300 above the line). Its studentized residual (3.97)** exceeds 3, so it is an unusual outlier in price. Its leverage (0.014 is about average and below 0.027, so it is not unusual (its age is typical). Its **Cook's distance (0.10)** is well below 1, so it is not influential: it does not noticeably change the fitted line.

(8)

```r
se.b1 <- sigma.hat / sqrt(SSX)
t.star <- qt(0.95, df = n - 2)
c(se = se.b1, t.star = t.star)
c(lower = b1 - t.star * se.b1, upper = b1 + t.star * se.b1)

confint(lm.fit, "age", level = 0.90)   # confirm
```

We are 90% confident that for each additional year of age, the mean price of this car model decreases by between $1,096 and $1,302.

(9)

```r
r^2                                  # square of the correlation
ss <- anova(lm.fit)$"Sum Sq"         # ANOVA: SSModel, SSE
ss[1] / (ss[1] + ss[2])              # SSModel / SSTotal
summary(lm.fit)$r.squared            # check
```

r² = (−0.8449)² = 1037.20 / 1452.80 = 0.714. About 71.4% of the variation in used car prices is explained by the linear relationship with age.

(10)

In each case, H₀: there is no linear relationship (ρ = 0, equivalently β₁ = 0) vs. Hₐ: there is a linear relationship (ρ ≠ 0, β₁ ≠ 0).

*Test for correlation:* $t = \dfrac{r\sqrt{n-2}}{\sqrt{1-r^2}} = \dfrac{-0.8449\sqrt{148}}{\sqrt{1-0.7139}} = -19.22$, reference distribution t with n − 2 = 148 df.

```r
t.r <- r * sqrt(n - 2) / sqrt(1 - r^2)
c(t = t.r, p.value = 2 * pt(-abs(t.r), df = n - 2))
cor.test(x, y)   # confirm
```

*Test for slope:* $t = \dfrac{\hat\beta_1 - 0}{SE(\hat\beta_1)} = \dfrac{-1.1987}{0.06237} = -19.22$, reference distribution t with 148 df.

```r
t.b <- b1 / se.b1
c(t = t.b, p.value = 2 * pt(-abs(t.b), df = n - 2))
summary(lm.fit)$coefficients
```

```r
MSModel <- SSModel / 1
MSE     <- SSE / (n - 2)
F.stat  <- MSModel / MSE
c(F = F.stat, p.value = pf(F.stat, 1, n - 2, lower.tail = FALSE))
anova(lm.fit)   # confirm
```

**Conclusion (all three tests):** the p-value is essentially 0 (about 5 × 10⁻⁴²; note t² = (−19.22)² = 369.4 = F), so we reject H₀. There is very strong evidence of a linear relationship between age and price.

(11)

```r
x0 <- 5
y.hat   <- b0 + b1 * x0
se.fit  <- sigma.hat * sqrt(1/n + (x0 - xbar)^2 / SSX)
se.pred <- sigma.hat * sqrt(1 + 1/n + (x0 - xbar)^2 / SSX)

y.hat
c(lower = y.hat - t.star * se.fit,  upper = y.hat + t.star * se.fit)    # 90% CI for mu_Y
c(lower = y.hat - t.star * se.pred, upper = y.hat + t.star * se.pred)   # 90% PI for Y

# confirm with predict(), as in ISLR
predict(lm.fit, data.frame(age = x0), interval = "confidence", level = 0.90)
predict(lm.fit, data.frame(age = x0), interval = "prediction", level = 0.90)
```

*Predicted value:* a 5-year-old car of this model is predicted to cost about $12,098.
*90% confidence interval for μ_Y (11.86, 12.34):* we are 90% confident that the **average** price of all 5-year-old cars of this model is between $11,860 and $12,337.
*90% prediction interval for an individual Y (9.31, 14.88):* we are 90% confident that the price of **one particular** 5-year-old car will be between $9,314 and $14,882.
The prediction interval is much wider because it accounts for the car-to-car variability around the mean (the "1 +" under the square root) in addition to the uncertainty in estimating the mean.

(12)

Setting the predicted price to 0: $0 = 18.0919 - 1.1987\,age \Rightarrow age = 18.0919/1.1987$.

```r
free.age <- -b0 / b1
free.age
max(cars$age)
```

The model says a car is free at about 15.1 years, which is beyond the oldest car in the data (13 years). Real used cars never become free (they keep scrap or parts value), and the residual plot suggested price levels off rather than falling linearly. So the line is only appropriate within the observed range of ages, and extrapolating past it is not trustworthy.

(13)

A simple linear regression of price on age explains about 71% of the variation in this used car's prices, with a highly significant negative slope of about −$1,200 per year and a typical error of about $1,680. The conditions are reasonably satisfied, with a mild curve in the residuals (depreciation slows with age), a slightly heavy right tail and somewhat more spread for the newest cars. The most unusual car is the 2-year-old listed at $22,000 (studentized residual ≈ 4), an outlier that is not influential. The oldest cars (age 12–13) have the highest leverage, and the extrapolated "free" age of about 15 years shows the model should not be used beyond the observed ages. Since mileage is also strongly related to price, a multiple regression with age and miles is a natural next step.
