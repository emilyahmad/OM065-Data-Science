```r
# seq(a,b) vector of integers between a & b
# seq(0,1,length=10) 10 equal spaced #s between

x=seq(1,10)
x
x=1:10 # splicing
xx=seq(-pi, pi, length=50)
```

```r
# 3d data, contour map (topographical)
# (vectorX, vY, matrix Z for each pair of (x,y) coords)
y=x
f=outer(x,y,function(x,y)cos(y)/(1+x^2))
contour(x,y,f)
contour(x,y,f,nlevels=45,add=T)
fa=(f-t(f))/2
contour(x,y,fa,nlevels=15)
```

```r
# image() heatmap - same as contour, color-coded plot based on z value
# persp() 3d plot, theta & phi for viewing angles
image(x,y,fa)
persp(x,y,fa)
persp(x,y,fa,theta=30,phi=20)
persp(x,y,fa,theta=30,phi=70)
persp(x,y,fa,theta=30,phi=40)
```
