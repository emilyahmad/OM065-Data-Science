(b)

<br>ln(odds of survival) = 1.0126 - 0.0386 * Age

```r
m_sex <- glm(Survive ~ Sex, data = donner, family = binomial)
m_age <- glm(Survive ~ Age, data = donner, family = binomial)
c(deviance(m_sex), deviance(m_age)); AIC(m_sex, m_age)
```

(c)<br>Age has a lower deviance so it is a slightly better predictor thang gender.

```r
exp(confint.default(m_sex))["Sex", ]
```

```r
donner$Age2 <- donner$Age^2
m_quad <- glm(Survive ~ Sex + Age + Age2, data = donner, family = binomial)
```

(a)<br>ln(odds) = 0.9425 - 1.2217 * Sex + 0.0731 * Age - 0.0023 * Age^2

```r
m_lin <- glm(Survive ~ Sex + Age, data = donner, family = binomial)
anova(m_lin, m_quad, test = "Chisq")
```

(b) The quadratic term for Age significantly imprpoved the model, as p < .05

```r
max(abs(residuals(m_quad, type = "pearson")))
```
