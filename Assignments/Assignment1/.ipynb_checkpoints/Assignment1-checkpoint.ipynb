```r
# next few problems use the variables we define as qualitative, so I'll create a vector
quant <- c("mpg", "displacement", "horsepower", "weight", "acceleration")

for (v in quant) {
  # (b) grab the range of each variable
  cat(v, ":", range(Auto[[v]]), "\n")
  # (c) mean
  cat(v, ":", mean(Auto[[v]]), "\n")
  # standard dev
  cat(v, ":", sd(Auto[[v]]), "\n\n")
}
```

```r
# (d) remove observations 10-85, get range, mean & std dev
Auto_reduced=Auto[-(10:85),]

for (v in quant) {
  # (b) grab the range of each variable
  cat(v, ":", range(Auto_reduced[[v]]), "\n")
  # (c) mean
  cat(v, ":", mean(Auto_reduced[[v]]), "\n")
  # standard dev
  cat(v, ":", sd(Auto_reduced[[v]]), "\n\n")
}
```

```r
predictors <- c("displacement", "horsepower", "weight", "acceleration")
par(mfrow = c(2, 2))   # 2x2 grid so all four plots appear together
for (v in predictors) {
  plot(Auto[[v]], Auto$mpg, xlab = v, ylab = "mpg", main = paste("mpg vs", v))
}
```

Based on these plots which graph mpg as the response and the other variables I classified as quantitative as the predictors, mpg appears to be correlated with the displacement, horsepower, and weight

Has 506 rows and 14 columns, where each row represents a town/suburb

```r
# (b)
plot(Boston$age, Boston$medv)
# Potential findings: as the age increases the median value slowly decreases

plot(Boston$age, Boston$tax)
# Potential findings: regardless of an increase in age, the full value property tax rate stays within the same range
```

```r
predictors <- setdiff(names(Boston), "crim")

for (v in predictors) {
  plot(Boston[[v]], Boston$crim, xlab = v, ylab = "Per capita crime rate")
}
```

There appears to be an association between per capita crime rate and the nitrogen oxide concentration (nox) and the proprtion of blacks (black)
