```r
library(ggplot2)

# 1. An empty plot: ggplot() only creates the blank canvas, no layers yet
ggplot(data = mpg)

# 2. Rows and columns in mpg
nrow(mpg)  # 234
ncol(mpg)  # 11
dim(mpg)

# 3. drv: type of drive train (f = front-wheel, r = rear-wheel, 4 = four-wheel)
?mpg
unique(mpg$drv)

# 4. Scatterplot of hwy vs cyl
ggplot(data = mpg) +
  geom_point(mapping = aes(x = cyl, y = hwy))

# 5. class vs drv: both variables are categorical, so points overlap
# at the few possible combinations and show nothing about frequency or trend
ggplot(data = mpg) +
  geom_point(mapping = aes(x = class, y = drv))
```

```r
# The color argument is inside aes(), so "blue" is treated as data: ggplot
# maps the string to a variable with one level and colors it with the
# default palette (red/salmon) and adds a legend.
# To set the color manually, move it outside aes():
ggplot(data = mpg) +
  geom_point(mapping = aes(x = displ, y = hwy), color = "blue")
```
