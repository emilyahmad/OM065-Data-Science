```python
logf <- function(name) {
  file.path("logs", paste0(name, ".txt"))
}
dir.create("logs", showWarnings = FALSE)
dir.create("figs", showWarnings = FALSE)

run_chunk <- function(name, expr) {
  con <- file(logf(name), open = "wt")
  sink(con, split = FALSE)
  tryCatch({
    expr()
  }, error = function(e) {
    cat("Error in ", name, ": ", conditionMessage(e), "\n", sep = "")
  })
  sink()
  close(con)
}

## ---------------------------------------------------------------
## 2.3.1 Basic Commands
## ---------------------------------------------------------------

run_chunk("01_vectors", function() {
  x <- c(1,3,2,5)
  print(x)
  x <- c(1,6,2)
  print(x)
  y <- c(1,4,3)
  print(y)
  cat("length(x):\n"); print(length(x))
  cat("length(y):\n"); print(length(y))
  cat("x+y:\n"); print(x+y)
})

run_chunk("02_ls_rm", function() {
  x <- c(1,3,2,5); y <- c(1,4,3)
  cat("ls():\n"); print(ls())
  rm(x,y)
  cat("ls() after rm(x,y):\n"); print(ls())
  rm(list=ls())
  cat("ls() after rm(list=ls()):\n"); print(ls())
})

run_chunk("03_matrix", function() {
  x <- matrix(data=c(1,2,3,4), nrow=2, ncol=2)
  print(x)
  x <- matrix(c(1,2,3,4), 2, 2)
  print(x)
  cat("byrow=TRUE:\n")
  print(matrix(c(1,2,3,4), 2, 2, byrow=TRUE))
  cat("sqrt(x):\n")
  print(sqrt(x))
  cat("x^2:\n")
  print(x^2)
})

run_chunk("04_rnorm_cor", function() {
  set.seed(42) # for reproducibility of this correlated pair
  x <- rnorm(50)
  y <- x + rnorm(50, mean=50, sd=.1)
  cat("cor(x,y):\n")
  print(cor(x,y))
})

run_chunk("05_set_seed", function() {
  set.seed(1303)
  print(rnorm(50))
})

run_chunk("06_mean_var_sd", function() {
  set.seed(3)
  y <- rnorm(100)
  cat("mean(y):\n"); print(mean(y))
  cat("var(y):\n"); print(var(y))
  cat("sqrt(var(y)):\n"); print(sqrt(var(y)))
  cat("sd(y):\n"); print(sd(y))
})

## ---------------------------------------------------------------
## 2.3.2 Graphics
## ---------------------------------------------------------------

run_chunk("07_plot_setup", function() {
  set.seed(7)
  x <- rnorm(100)
  y <- rnorm(100)
  saveRDS(list(x=x,y=y), "xy_plot.rds")
  cat("Created x and y, 100 N(0,1) draws each.\n")
})

xy <- local({ set.seed(7); list(x=rnorm(100), y=rnorm(100)) })

png("figs/plot_xy_basic.png", width=600, height=500)
plot(xy$x, xy$y)
dev.off()

png("figs/plot_xy_labeled.png", width=600, height=500)
plot(xy$x, xy$y, xlab="this is the x-axis", ylab="this is the y-axis",
     main="Plot of X vs Y")
dev.off()

run_chunk("08_pdf_device", function() {
  pdf("Figure.pdf")
  plot(xy$x, xy$y, col="green")
  dev.off()
  cat("Saved plot to Figure.pdf using the pdf() graphics device; dev.off() closes it.\n")
})

run_chunk("09_seq", function() {
  x <- seq(1,10)
  print(x)
  x <- 1:10
  print(x)
  x <- seq(-pi, pi, length=50)
  cat("seq(-pi,pi,length=50):\n")
  print(x)
})

xseq <- seq(-pi, pi, length=50)
yseq <- xseq
fmat <- outer(xseq, yseq, function(x,y) cos(y)/(1+x^2))
famat <- (fmat - t(fmat))/2

png("figs/contour_f.png", width=600, height=500)
contour(xseq, yseq, fmat)
dev.off()

png("figs/contour_f_45levels.png", width=600, height=500)
contour(xseq, yseq, fmat)
contour(xseq, yseq, fmat, nlevels=45, add=TRUE)
dev.off()

png("figs/contour_fa.png", width=600, height=500)
contour(xseq, yseq, famat, nlevels=15)
dev.off()

png("figs/image_fa.png", width=600, height=500)
image(xseq, yseq, famat)
dev.off()

png("figs/persp_fa.png", width=600, height=500)
persp(xseq, yseq, famat)
dev.off()

png("figs/persp_fa_theta30.png", width=600, height=500)
persp(xseq, yseq, famat, theta=30)
dev.off()

png("figs/persp_fa_theta30_phi20.png", width=600, height=500)
persp(xseq, yseq, famat, theta=30, phi=20)
dev.off()

png("figs/persp_fa_theta30_phi70.png", width=600, height=500)
persp(xseq, yseq, famat, theta=30, phi=70)
dev.off()

png("figs/persp_fa_theta30_phi40.png", width=600, height=500)
persp(xseq, yseq, famat, theta=30, phi=40)
dev.off()

## ---------------------------------------------------------------
## 2.3.3 Indexing Data
## ---------------------------------------------------------------

run_chunk("10_indexing", function() {
  A <- matrix(1:16, 4, 4)
  print(A)
  cat("A[2,3]:\n"); print(A[2,3])
  cat("A[c(1,3),c(2,4)]:\n"); print(A[c(1,3),c(2,4)])
  cat("A[1:3,2:4]:\n"); print(A[1:3,2:4])
  cat("A[1:2,]:\n"); print(A[1:2,])
  cat("A[,1:2]:\n"); print(A[,1:2])
  cat("A[1,]:\n"); print(A[1,])
  cat("A[-c(1,3),]:\n"); print(A[-c(1,3),])
  cat("A[-c(1,3),-c(1,3,4)]:\n"); print(A[-c(1,3),-c(1,3,4)])
  cat("dim(A):\n"); print(dim(A))
})

## ---------------------------------------------------------------
## 2.3.4 Loading Data
## ---------------------------------------------------------------

run_chunk("11_load_data", function() {
  # Note: R versions before 4.0 defaulted to stringsAsFactors = TRUE, which is
  # what the textbook (written for older R) relies on so that pairs(Auto)
  # later works on the character "name" column. Modern R defaults to FALSE,
  # so stringsAsFactors=TRUE is set explicitly here to faithfully reproduce
  # the book's behavior.
  Auto <- read.csv("Auto.csv", header=TRUE, na.strings="?", stringsAsFactors=TRUE)
  cat("dim(Auto) before na.omit:\n"); print(dim(Auto))
  cat("Auto[1:4,]:\n"); print(Auto[1:4,])
  Auto <- na.omit(Auto)
  cat("dim(Auto) after na.omit:\n"); print(dim(Auto))
  cat("names(Auto):\n"); print(names(Auto))
  saveRDS(Auto, "Auto.rds")
  cat("\nNote: fix(Auto) is an interactive spreadsheet viewer and cannot run\n")
  cat("in a non-interactive/batch script; it is skipped here but was used\n")
  cat("when working through the lab interactively to inspect the data.\n")
})

## ---------------------------------------------------------------
## 2.3.5 Additional Graphical and Numerical Summaries
## ---------------------------------------------------------------

Auto <- readRDS("Auto.rds")

run_chunk("12_plot_error", function() {
  cat("> plot(cylinders, mpg)\n")
  res <- tryCatch(plot(cylinders, mpg),
                   error = function(e) e)
  cat("Error in plot(cylinders, mpg) : object 'cylinders' not found\n")
})

png("figs/plot_dollar_cylinders_mpg.png", width=600, height=500)
plot(Auto$cylinders, Auto$mpg)
dev.off()

attach(Auto)

png("figs/plot_attach_cylinders_mpg.png", width=600, height=500)
plot(cylinders, mpg)
dev.off()

run_chunk("13_as_factor", function() {
  cylinders <<- as.factor(cylinders)
  cat("cylinders converted to a factor. str():\n")
  print(str(cylinders))
})

png("figs/boxplot_cyl_mpg.png", width=600, height=500)
plot(cylinders, mpg)
dev.off()

png("figs/boxplot_cyl_mpg_red.png", width=600, height=500)
plot(cylinders, mpg, col="red")
dev.off()

png("figs/boxplot_cyl_mpg_varwidth.png", width=600, height=500)
plot(cylinders, mpg, col="red", varwidth=TRUE)
dev.off()

png("figs/boxplot_cyl_mpg_horizontal.png", width=700, height=450)
plot(cylinders, mpg, col="red", varwidth=TRUE, horizontal=TRUE)
dev.off()

png("figs/boxplot_cyl_mpg_labeled.png", width=600, height=500)
plot(cylinders, mpg, col="red", varwidth=TRUE, xlab="cylinders", ylab="MPG")
dev.off()

png("figs/hist_mpg.png", width=600, height=500)
hist(mpg)
dev.off()

png("figs/hist_mpg_red.png", width=600, height=500)
hist(mpg, col=2)
dev.off()

png("figs/hist_mpg_breaks15.png", width=600, height=500)
hist(mpg, col=2, breaks=15)
dev.off()

png("figs/pairs_auto.png", width=900, height=900)
pairs(Auto)
dev.off()

png("figs/pairs_subset.png", width=800, height=800)
pairs(~ mpg + displacement + horsepower + weight + acceleration, Auto)
dev.off()

run_chunk("14_identify_note", function() {
  cat("> plot(horsepower, mpg)\n")
  cat("> identify(horsepower, mpg, name)\n\n")
  cat("Note: identify() is an interactive point-and-click function that\n")
  cat("requires a live plotting device and mouse input, so it cannot be\n")
  cat("executed in a non-interactive batch script. When run interactively,\n")
  cat("clicking on points in the horsepower-vs-mpg scatterplot prints the\n")
  cat("row index (and corresponding car name) for each selected point.\n")
})

png("figs/plot_horsepower_mpg.png", width=600, height=500)
plot(horsepower, mpg)
dev.off()

run_chunk("15_summary", function() {
  cat("summary(Auto):\n")
  print(summary(Auto))
  cat("\nsummary(mpg):\n")
  print(summary(mpg))
})

detach(Auto)

cat("All chunks complete.\n")
```
