```python
# College dataset
# (a) Read in the data
college <- read.csv("College.csv")
dim(college)
names(college)

# (b) Fix row names
rownames(college) <- college[, 1]
fix(college)
college <- college[, -1]
fix(college)

# (c) i. Numerical summary
summary(college)

# (c) ii. Scatterplot matrix of first 10 variables
pairs(college[, 1:10])

# (c) iii. Boxplot of Outstate vs Private
plot(as.factor(college$Private), college$Outstate,
     xlab = "Private", ylab = "Out-of-State Tuition ($)")

# (c) iv. Create the Elite variable
Elite <- rep("No", nrow(college))
Elite[college$Top10perc > 50] <- "Yes"
Elite <- as.factor(Elite)
college <- data.frame(college, Elite)
summary(college$Elite)

plot(college$Elite, college$Outstate,
     xlab = "Elite", ylab = "Out-of-State Tuition ($)")

# (c) v. Histograms in a 2x2 grid
par(mfrow = c(2, 2))
hist(college$Apps, breaks = 10, main = "Applications", xlab = "Apps")
hist(college$Enroll, breaks = 20, main = "Enrollment", xlab = "Enroll")
hist(college$Outstate, breaks = 15, main = "Out-of-State Tuition", xlab = "Outstate")
hist(college$Grad.Rate, breaks = 25, main = "Graduation Rate", xlab = "Grad.Rate")
par(mfrow = c(1, 1))

# (c) vi. Further exploration
acc.rate <- college$Accept / college$Apps
summary(acc.rate)
rownames(college)[which.min(acc.rate)]   # most selective school
min(acc.rate)
rownames(college)[which.max(acc.rate)]   # least selective school
max(acc.rate)

which(college$Grad.Rate > 100)           # data quality check
college$Grad.Rate[college$Grad.Rate > 100]

cor(college$Outstate, college$Expend)
cor(college$PhD, college$Grad.Rate)
table(college$Private, college$Elite)

plot(college$Expend, college$Outstate,
     xlab = "Expenditure per Student", ylab = "Out-of-State Tuition",
     col = ifelse(college$Private == "Yes", "orange", "gray40"), pch = 16)
legend("bottomright", legend = c("Private", "Public"),
       col = c("orange", "gray40"), pch = 16)


## ============================================================
## Exercise 9: The Auto Data Set
## ============================================================

Auto <- read.csv("Auto.csv", header = TRUE, na.strings = "?")
Auto <- na.omit(Auto)
dim(Auto)
str(Auto)

# (a) Quantitative vs qualitative -- inspect with str(), then note:
# Quantitative: mpg, displacement, horsepower, weight, acceleration
# Qualitative:  cylinders, year, origin, name

quant.vars <- c("mpg", "displacement", "horsepower", "weight", "acceleration")

# (b) Range of each quantitative predictor
sapply(Auto[, quant.vars], range)

# (c) Mean and standard deviation of each quantitative predictor
sapply(Auto[, quant.vars], mean)
sapply(Auto[, quant.vars], sd)

# (d) Remove observations 10-85, redo range/mean/sd
Auto2 <- Auto[-(10:85), ]
dim(Auto2)
sapply(Auto2[, quant.vars], range)
sapply(Auto2[, quant.vars], mean)
sapply(Auto2[, quant.vars], sd)

# (e) Graphical investigation
pairs(Auto[, c(quant.vars, "cylinders", "year", "origin")])

plot(Auto$weight, Auto$mpg, xlab = "Weight", ylab = "MPG")
plot(Auto$horsepower, Auto$mpg, xlab = "Horsepower", ylab = "MPG")

plot(as.factor(Auto$origin), Auto$mpg, xlab = "Origin", ylab = "MPG")
plot(as.factor(Auto$cylinders), Auto$mpg, xlab = "Cylinders", ylab = "MPG")

# (f) Predicting mpg -- correlations
sort(cor(Auto[, c(quant.vars, "year")])[, "mpg"])


## ============================================================
## Exercise 10: The Boston Housing Data Set
## ============================================================

library(MASS)
dim(Boston)
names(Boston)
?Boston

# (b) Pairwise scatterplots
pairs(Boston)

# (c) Predictors associated with crime rate
sort(cor(Boston)[, "crim"])

plot(Boston$rad, Boston$crim, xlab = "rad", ylab = "crim")
plot(Boston$tax, Boston$crim, xlab = "tax", ylab = "crim")

# (d) Suburbs with high crime, tax, ptratio
range(Boston$crim); summary(Boston$crim)
range(Boston$tax); summary(Boston$tax)
range(Boston$ptratio); summary(Boston$ptratio)

Boston[order(-Boston$crim), ][1:5, c("crim", "tax", "ptratio")]

hist(Boston$crim, breaks = 40, main = "Crime Rate")
hist(Boston$tax, breaks = 20, main = "Tax Rate")
hist(Boston$ptratio, breaks = 20, main = "Pupil-Teacher Ratio")

# (e) Suburbs bounding the Charles river
sum(Boston$chas == 1)

# (f) Median pupil-teacher ratio
median(Boston$ptratio)

# (g) Suburb with lowest median home value
which.min(Boston$medv)
Boston[which.min(Boston$medv), ]
summary(Boston)   # for comparison

# (h) Suburbs with many rooms per dwelling
sum(Boston$rm > 7)
sum(Boston$rm > 8)
summary(Boston[Boston$rm > 8, ])
```
