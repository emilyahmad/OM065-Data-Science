## Setup and data cleaning
- `Age` is missing for 177 training passengers. Missing ages are filled in with the median age for the passenger's title group (Mr, Mrs, Miss, Master, Other), using **training data only**.
- `Embarked` is missing for 2 passengers; they are set to the most common port, "S".
- Test file: `testWsurvived.csv` (418 passengers). The 20 rows with no `Survived` value are dropped, which leaves 398. The 1 missing `Fare` is filled with the 3rd-class median fare from training.

```r
library(car)

train <- read.csv("TitanicTrain.csv", na.strings = c("", "NA"))
test  <- read.csv("testWsurvived.csv", na.strings = c("", "NA"))

# Extract title from Name (e.g. "Braund, Mr. Owen Harris" -> "Mr")
train$Title <- sub(".*, *([^.]+)\\..*", "\\1", train$Name)
test$Title  <- sub(".*, *([^.]+)\\..*", "\\1", test$Name)
table(train$Title)

grp <- function(t) ifelse(t %in% c("Mr","Mrs","Miss","Master"), t,
                   ifelse(t %in% c("Mlle","Ms"), "Miss",
                   ifelse(t == "Mme", "Mrs", "Other")))
train$TitleGrp <- grp(train$Title)
test$TitleGrp  <- grp(test$Title)

# Impute missing Age with title-group median (from training data)
age_med <- tapply(train$Age, train$TitleGrp, median, na.rm = TRUE)
age_med
train$Age[is.na(train$Age)] <- age_med[train$TitleGrp[is.na(train$Age)]]
test$Age[is.na(test$Age)]   <- age_med[test$TitleGrp[is.na(test$Age)]]

train$Embarked[is.na(train$Embarked)] <- "S"
test$Fare[is.na(test$Fare)] <- median(train$Fare[train$Pclass == 3])
test <- test[!is.na(test$Survived), ]   # drop rows with unknown survival

# Factors (test uses the same levels as train)
for (v in c("Pclass", "Sex", "Embarked")) {
  train[[v]] <- factor(train[[v]])
  test[[v]]  <- factor(test[[v]], levels = levels(train[[v]]))
}
c(train = nrow(train), test = nrow(test))
```

## Part 1. Build an initial model
Survival is a 0/1 response, so the model is a **logistic regression**. Model selection approach:
1. Fit the full model with all usable predictors.
2. Run backward, forward, and stepwise selection by AIC, plus backward elimination by BIC.
3. Compare the candidate models by AIC/BIC and with drop-in-deviance (likelihood-ratio) tests.

```r
full <- glm(Survived ~ Pclass + Sex + Age + SibSp + Parch + Fare + Embarked,
            family = binomial, data = train)
summary(full)
```

```r
null <- glm(Survived ~ 1, family = binomial, data = train)

back <- step(full, direction = "backward", trace = 0)
fwd  <- step(null, scope = formula(full), direction = "forward", trace = 0)
both <- step(null, scope = formula(full), direction = "both", trace = 0)
bic  <- step(full, direction = "backward", k = log(nrow(train)), trace = 0)

formula(back); formula(fwd); formula(both); formula(bic)

mods <- list(full = full, backAIC = back, fwdAIC = fwd, stepAIC = both, backBIC = bic)
data.frame(AIC = sapply(mods, AIC), BIC = sapply(mods, BIC))
```

```r
# The AIC methods keep Embarked; BIC drops it. Use likelihood-ratio tests to decide.
model1 <- glm(Survived ~ Pclass + Sex + Age + SibSp, family = binomial, data = train)

anova(model1, back, test = "Chisq")   # does adding Embarked help?
anova(model1, full, test = "Chisq")   # do Parch + Fare + Embarked together help?
anova(update(model1, . ~ . - SibSp), model1, test = "Chisq")  # is SibSp needed?
```

**Initial model:** `Survived ~ Pclass + Sex + Age + SibSp`

```r
summary(model1)
vif(model1)
```

*Write-up notes:* comment on which predictors are significant at the 5% level, and on the VIFs. Values near 1 mean the predictors are not collinear; the usual concern starts around 5.

## Part 2. Residual analysis
Criteria for "unusual" points (n = 891, p = 6 coefficients):
- Studentized residual: |r| > 3
- Leverage: h > 3p/n
- Cook's D: > 0.5 is influential (4/n is shown only as a milder screen)

```r
par(mfrow = c(2, 2)); plot(model1); par(mfrow = c(1, 1))
```

```r
residualPlots(model1)        # Pearson residuals vs each predictor / linear predictor
marginalModelPlots(model1)   # empirical logit fit vs model fit (checks linearity in the logit)
```

```r
sres <- rstudent(model1)
lev  <- hatvalues(model1)
cd   <- cooks.distance(model1)
p <- length(coef(model1)); n <- nrow(train)

big_sres <- which(abs(sres) > 3);  big_sres
high_lev <- which(lev > 3 * p / n); high_lev
which(cd > 4 / n)    # mild screen
which(cd > 0.5)      # influential

influencePlot(model1)
```

```r
# Look at the flagged cases
unusual <- union(big_sres, high_lev)
cbind(train[unusual, c("PassengerId","Survived","Pclass","Sex","Age","SibSp")],
      sres = round(sres[unusual], 2), lev = round(lev[unusual], 3), cooksD = round(cd[unusual], 4))
```

```r
# Sensitivity check: refit without the flagged cases and compare coefficients
model2 <- update(model1, data = train[-unusual, ])
summary(model2)
round(cbind(model1 = coef(model1), model2 = coef(model2)), 4)
```

*Write-up notes:* no case has Cook's D > 0.5, so no single passenger drives the fit. The large studentized residuals are mostly "surprising" outcomes, such as 3rd-class adult men who survived. Removing them just exaggerates the coefficients, so the final model keeps the full data (`model1`).

## Part 3. Cross-validation

```r
finalTrain <- model1

predQ  <- predict(finalTrain, newdata = test, type = "response")   # predicted P(survive)
residQ <- test$Survived - predQ

# (a) holdout residuals vs training residuals
c(mean = mean(residQ), sd = sd(residQ))
resTrain <- residuals(finalTrain, type = "response")
c(mean = mean(resTrain), sd = sd(resTrain))
```

```r
# (b) poorly predicted holdout cases (|residual / sd| > 2)
stdResid <- residQ / sd(residQ)
poor <- which(abs(stdResid) > 2)
data.frame(holdoutRow = poor, test[poor, c("PassengerId","Survived","Pclass","Sex","Age","SibSp")],
           pred = round(predQ[poor], 3), stdResid = round(stdResid[poor], 2))
```

```r
# (c) cross-validation correlation and shrinkage
cvCor   <- cor(predQ, test$Survived)
R2train <- cor(fitted(finalTrain), train$Survived)^2   # training R^2 (squared corr of fitted vs actual)
c(cvCor = cvCor, cvR2 = cvCor^2, trainR2 = R2train, shrinkage = R2train - cvCor^2)

# Classification accuracy on the holdout at a 0.5 cutoff
conf <- table(Actual = test$Survived, Predicted = as.numeric(predQ > 0.5)); conf
sum(diag(conf)) / sum(conf)
```

## Part 4. Final model

```r
final <- model1
summary(final)
```

```r
# Final prediction equation (copy/paste-able R)
b <- coef(final)
cat(sprintf(paste0("logit <- %.6f + (%.6f)*(Pclass==2) + (%.6f)*(Pclass==3) + (%.6f)*(Sex==\"male\") + (%.6f)*Age + (%.6f)*SibSp\n",
                   "probSurvive <- exp(logit) / (1 + exp(logit))\n"),
            b[1], b[2], b[3], b[4], b[5], b[6]))
```

```r
# Check: the hand-written equation matches predict()
logit <- with(test, b[1] + b[2]*(Pclass==2) + b[3]*(Pclass==3) + b[4]*(Sex=="male") + b[5]*Age + b[6]*SibSp)
all.equal(as.numeric(exp(logit)/(1+exp(logit))), as.numeric(predQ))
```

## Extra Credit
**EC1:** the test file with true survival is `testWsurvived.csv`. Cleaning steps are in the Setup cell: dropped 20 rows with missing `Survived`, imputed `Age` by title-group median, and imputed 1 missing `Fare`.

```r
# EC2: Precision and Recall (0.5 cutoff)
pr <- function(actual, prob, cut = 0.5) {
  pred <- as.numeric(prob > cut)
  tp <- sum(pred == 1 & actual == 1); fp <- sum(pred == 1 & actual == 0); fn <- sum(pred == 0 & actual == 1)
  c(Precision = tp / (tp + fp), Recall = tp / (tp + fn), Accuracy = mean(pred == actual))
}
rbind(Train = pr(train$Survived, fitted(final)), Test = pr(test$Survived, predQ))
```

```r
# EC3: new candidate predictors
train$Military <- as.numeric(train$Title %in% c("Capt", "Col", "Major"))
train$Noble    <- as.numeric(train$Title %in% c("Countess", "the Countess", "Lady", "Sir", "Don", "Dona", "Jonkheer"))
train$Child    <- as.numeric(train$Age < 13)
train$Deck     <- ifelse(is.na(train$Cabin), "Unknown", substr(train$Cabin, 1, 1))
table(train$Military); table(train$Noble); table(train$Child); table(train$Deck)

for (v in c("Military", "Noble", "Child")) {
  m <- update(final, as.formula(paste(". ~ . +", v)))
  cat("\n=====", v, "=====\n")
  print(summary(m)$coefficients[v, ])
  print(anova(final, m, test = "Chisq"))
}
mDeck <- update(final, . ~ . + factor(Deck))
anova(final, mDeck, test = "Chisq")
```
