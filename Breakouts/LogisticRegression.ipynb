what we should focus on to accomodate data that's more varied than actual binomial dist
Dispersion parameter for binomial family taken to be 1

dispersion - variance parameter of the binomial distribution (in any distribution, can calculate mean, std dev), np(1-p)

mean = np (# trials * proportion of success)
std dev (sigma^2) = np(1-p)

to make more dispersed data, put parameter in front & multiply by a coeffecient
phi(np)(1-p) > typical variance of binomial distribution
or sigma^2 = phi(np)(1-p) > 1 makes it narrower/smaller dist
simple way to stretch/compress dist
*just another coeffecient we can alter
(quasi binomial)

R can fit different values of phi to best fit the training sample

not parameterization, didn't express something in terms of parameter = chose parameters to express it in
parameter already existed, "dispersion variable" was just = 1, considered a constant

perform maximization for log(likelihood) - what fit's best this pattern, could make that a variable as well. Additional parameter = additional flexibility.
