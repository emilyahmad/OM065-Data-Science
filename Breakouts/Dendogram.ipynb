```r
#install.packages("cluster")
library(ggplot2)
library(repr)
par(mfrow=c(1,3))
# Set the parameters for the plot sizes for all plots below
options(repr.plot.width=9, repr.plot.height=6)
```

```r
# load the data
data(iris)
```

```r
d <- dist(iris[, 1:4])
hcl <- hclust(d, method = "complete")
hcl
plot(hcl)
```

```r
hcd = as.dendrogram(hcl)
```

```r
plot(cut(hcd, h=1.5)$upper, main="Cut to specific height", hang = -1)
```

```r
km = kmeans(iris[,1:4], centers = 3, nstart = 10)
hcl <- hclust(dist(iris[,1:4]))
```

```r
table(km$cluster, cutree(hcl, k=3))
```
