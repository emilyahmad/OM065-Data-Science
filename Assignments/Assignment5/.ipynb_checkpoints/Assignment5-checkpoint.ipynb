```r
# Exercise 10 (a) Numerical and graphical summaries
library(ISLR)
library(MASS)
library(class)

names(Weekly)
dim(Weekly)
summary(Weekly)

# Pairwise correlations among the numeric variables (Direction is column 9)
cor(Weekly[, -9])

# Graphical summaries
pairs(Weekly)
plot(Weekly$Volume, ylab = "Volume (billions of shares)", main = "Volume over time (week index)")
boxplot(Today ~ Direction, data = Weekly, main = "Today's return by Direction")
table(Weekly$Direction)
```

```r
# (b) Logistic regression on the full data set
glm.fit <- glm(Direction ~ Lag1 + Lag2 + Lag3 + Lag4 + Lag5 + Volume,
               data = Weekly, family = binomial)
summary(glm.fit)
```

```r
# (c) Confusion matrix and overall fraction of correct predictions
glm.probs <- predict(glm.fit, type = "response")
glm.pred.full <- ifelse(glm.probs > 0.5, "Up", "Down")
table(glm.pred.full, Weekly$Direction)
mean(glm.pred.full == Weekly$Direction)
```

```r
# (d) Logistic regression, train 1990-2008, Lag2 only; evaluate on 2009-2010
train <- Weekly$Year < 2009
Weekly.test <- Weekly[!train, ]
Direction.train <- Weekly$Direction[train]
Direction.test  <- Weekly$Direction[!train]

glm.fit2 <- glm(Direction ~ Lag2, data = Weekly, family = binomial, subset = train)
glm.probs2 <- predict(glm.fit2, Weekly.test, type = "response")
glm.pred <- ifelse(glm.probs2 > 0.5, "Up", "Down")
table(glm.pred, Direction.test)
mean(glm.pred == Direction.test)
```

```r
# (e) LDA with Lag2 only
lda.fit <- lda(Direction ~ Lag2, data = Weekly, subset = train)
lda.fit
lda.pred <- predict(lda.fit, Weekly.test)
table(lda.pred$class, Direction.test)
mean(lda.pred$class == Direction.test)
```

```r
# (f) QDA with Lag2 only
qda.fit <- qda(Direction ~ Lag2, data = Weekly, subset = train)
qda.fit
qda.pred <- predict(qda.fit, Weekly.test)
table(qda.pred$class, Direction.test)
mean(qda.pred$class == Direction.test)
```

```r
# (g) KNN with K = 1 (Lag2 only)
train.X <- as.matrix(Weekly$Lag2[train])
test.X  <- as.matrix(Weekly$Lag2[!train])

set.seed(1)
knn.pred <- knn(train.X, test.X, Direction.train, k = 1)
table(knn.pred, Direction.test)
mean(knn.pred == Direction.test)
```

```r
# (h) Compare test accuracy across methods (Lag2 only, trained 1990-2008)
results <- data.frame(
  Method = c("Logistic", "LDA", "QDA", "KNN (K=1)"),
  Accuracy = c(mean(glm.pred == Direction.test),
               mean(lda.pred$class == Direction.test),
               mean(qda.pred$class == Direction.test),
               mean(knn.pred == Direction.test))
)
results
```

```r
# (i) Experiments on the held-out data (2009-2010)
acc <- function(pred) mean(pred == Direction.test)

# Logistic regression: Lag1 * Lag2 interaction
glm.i <- glm(Direction ~ Lag1 * Lag2, data = Weekly, family = binomial, subset = train)
pred.i <- ifelse(predict(glm.i, Weekly.test, type = "response") > 0.5, "Up", "Down")
table(pred.i, Direction.test); acc(pred.i)

# LDA: Lag1 + Lag2
lda.i <- lda(Direction ~ Lag1 + Lag2, data = Weekly, subset = train)
pred.i <- predict(lda.i, Weekly.test)$class
table(pred.i, Direction.test); acc(pred.i)

# QDA: Lag2 plus a squared term
qda.i <- qda(Direction ~ Lag2 + I(Lag2^2), data = Weekly, subset = train)
pred.i <- predict(qda.i, Weekly.test)$class
table(pred.i, Direction.test); acc(pred.i)

# KNN: try many values of K using Lag1 and Lag2
train.X2 <- cbind(Weekly$Lag1, Weekly$Lag2)[train, ]
test.X2  <- cbind(Weekly$Lag1, Weekly$Lag2)[!train, ]
ks <- 1:100
knn.acc <- sapply(ks, function(k) {
  set.seed(1)
  acc(knn(train.X2, test.X2, Direction.train, k = k))
})
plot(ks, knn.acc, type = "l", xlab = "K", ylab = "Test accuracy")
best.k <- ks[which.max(knn.acc)]
best.k; max(knn.acc)
set.seed(1)
table(knn(train.X2, test.X2, Direction.train, k = best.k), Direction.test)
```
