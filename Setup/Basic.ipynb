### Repeating R intro

```r
x <- c(1, 3, 4, 6, 5, 6)
x
```

```r
mean(x)
```

```r
sd(x)
```

```r
y <- matrix(x, nrow=3)
y
```

```r
apply(y,1,sum)
apply(y,2,sum)
```

```r
z <- matrix(x, nrow=3, byrow=TRUE)
```

```r
z
```

```r
name <- c("Emily", "Tim", "Jeff")
my.mat <- cbind(name, z)
is.matrix(my.mat)
is.data.frame(my.mat)
my.mat
my.mat[,2]
```

```r
my.data <- data.frame(name, z)
is.data.frame(my.data)
mean(my.data[,2])
```

```r
names(my.data) <- c("Name","Teeth", "Crawl")
my.data
mean(my.data$Teeth)
```
