```r
library(tidyverse)
data(mpg)
```

```r
ggplot(data = mpg, mapping = aes(x = displ, y = hwy)) +
  geom_point() +
  labs(
    title = "Engine Displacement vs. Highway MPG",
    x = "Engine Displacement (litres)",
    y = "Highway MPG"
  )
```

```r
ggplot(data = mpg, mapping = aes(x = displ, y = hwy, color = class, size = cyl)) +
  geom_point(alpha = 0.7) +
  labs(
    title = "Engine Displacement vs. Highway MPG",
    x = "Engine Displacement (litres)",
    y = "Highway MPG",
    color = "Vehicle Class",
    size = "Number of Cylinders"
  ) +
  theme_minimal()
```

```r
library(ggplot2)

ggplot(data = mpg, mapping = aes(x = displ, y = hwy, size = class)) +
  geom_point(alpha = 0.6) +
  labs(
    title = "Engine Displacement vs. Highway MPG",
    x = "Engine Displacement (litres)",
    y = "Highway MPG",
    size = "Vehicle Class"
  ) +
  theme_minimal()
```

```r
ggplot(data = mpg, mapping = aes(x = displ, y = hwy, shape = class)) +
  geom_point(size = 3, alpha = 0.8) +
  theme_minimal()
```

```r
ggplot(data = mpg, mapping = aes(x = displ, y = hwy, color = class)) +
  geom_point(size = 2.5, show.legend = FALSE) +
  facet_wrap(~class) +
  labs(
    title = "Displacement vs. Highway MPG by Vehicle Class",
    x = "Engine Displacement (litres)",
    y = "Highway MPG"
  ) +
  theme_light()
```

```r
ggplot(data = mpg, mapping = aes(x = displ, y = hwy, color = as.factor(cyl))) +
  geom_point(size = 2.5, show.legend = FALSE) + 
  facet_wrap(~cyl) +
  labs(
    title = "Displacement vs. Highway MPG by Cylinder Number",
    x = "Engine Displacement (litres)",
    y = "Highway MPG"
  ) +
  theme_light()
```

```r
ggplot(mpg) + 
  geom_bar(aes(y = class)) + 
  facet_wrap(~year) + 
  theme_minimal()
```
